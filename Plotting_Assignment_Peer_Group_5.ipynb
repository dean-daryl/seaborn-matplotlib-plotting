```python

numerical_df = df.select_dtypes(include=['number'])
corr_matrix = numerical_df.corr()

sns.heatmap(corr_matrix,
            cbar=True,
            annot=False,
            cmap = "coolwarm",
            fmt='.2f'
)
plt.title("Heatmap Correlation")

plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
'''
The left to right diagonal consists of high correlation (Self-correlation). It's expected and confirms each value is getting compared by itself.
'''

'''
STRONG POSITIVE CORRELATIONS:

1. comb(combined fuel efficiency) and highway: combined fuel efficiency typically depends on highway performance
2. ghgScore(Greenhouse gas score) and feScore(Fuel Efficiency Score): better fuel efficiency corresponds to lower greenhouse gas emissions
'''

'''
STRONG NEGATIVE CORRELATIONS:

1. co2(Carbon Dioxide) and ghgScore(Greenhouse gas score):higher co2 emissions result in a low ghg score.
2. co2(Greenhouse gas score) and feScore(Fuel Efficiency Score): higher co2 emissions result in a low fuel efficiency.
'''

'''
WEAK OR NO CORRELATIONS:

1. year and other variables: The year variable has little correlation with most other variables. This means that the year of the vehicle doesn't influence other features in the dataset.
'''

'''
STORY THE DATA TELLS:
 1. Vehicles with high fuel efficiency(feScore, ghgScore) emit less co2 and have small engine displacements(displ).
 2. Large engines(displ) and vehicles with more cylinders(cylinders) are not very fuel-efficient(feScore) and emit more greenhose gases(ghgScore).
 3. Improving vehicle fuel efficiency(feScore) is important for less greenhouse gas emissions(ghgScore).
 4. The year of the vehicle doesn't impact any of the other metrics. According to our analysis, we concluded new vehicles have not improved performace of emissions in general.

'''

```
